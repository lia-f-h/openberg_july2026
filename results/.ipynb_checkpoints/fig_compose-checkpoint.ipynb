```python
from PIL import Image
path ='/home/jovyan/work/tutorials/sources/OpenDrift/openberg_july2026/results/'
```

```python
f1="analysis_polarplot_iceberg2017b_P1.png"
f2="analysis_polarplot_iceberg2018b_Pfull.png"
f3="analysis_polarplot_iceberg2025w_P3.png"
f4="analysis_polarplot_iceberg2017b_P2.png"
f5="analysis_polarplot_legend_ncol5.png"
note = 'bic'

fig1 = Image.open(path+f1)
fig2 = Image.open(path+f2)
fig3 = Image.open(path+f3)
fig4 = Image.open(path+f4)
fig5 = Image.open(path+f5)

# top row
w = fig1.width + fig2.width + fig3.width +fig4.width
h = max(fig1.height, fig2.height,fig3.height, fig4.height)

top = Image.new("RGB", (w, h), "white")
top.paste(fig1, (0, 0))
top.paste(fig2, (fig1.width, 0))
top.paste(fig3, (fig1.width+fig2.width, 0))
top.paste(fig4, (fig1.width+fig2.width+fig3.width, 0))
# top.save("./polar_sim_%s.png"%note)

# 2nd row
final = Image.new("RGB", (w, h + fig5.height), "white")
final.paste(top, (0, 0))
final.paste(fig5, ((w - fig5.width)//2, h))

final.save("./polar_sim_%s.png"%note)
```
