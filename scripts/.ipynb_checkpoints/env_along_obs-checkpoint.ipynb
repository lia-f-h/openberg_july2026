```python
#simulation list
# file_l = glob.glob("./results/%s/*.nc"%ib)
file_l = glob.glob("./results/%s*.nc"%ib)
file_l = [f for f in file_l if all(ff in f for ff in sim_l_contain)]
file_l = [f for f in file_l if all(ff not in f for ff in sim_l_notcontain)]
print(file_l)
```

#### Cmems along obs

```python
# add env along obs
out = out.assign_coords(cmems_id=("cmems_id", cmems_l))#[env_dict[cm] if cm!='topaz5-ensemble' else env_dict[cm][0] for cm in cmems_l]))
    
for v in ['x_sea_water_velocity','y_sea_water_velocity',"sea_ice_area_fraction","sea_ice_x_velocity","sea_ice_y_velocity","x_wind","y_wind"]:
    out[v] = (('time_obs','cmems_id'), np.full((obs.time.data.size,len(cmems_l)),np.nan))
```

```python
out
```

```python
for cmems_id in cmems_l:
    
    if cmems_id in ['windglophyre','windglophynrt']:variables=['x_wind','y_wind']
    elif cmems_id in ['topaz6','glophyanfcH']: variables = ['x_sea_water_velocity','y_sea_water_velocity']
    elif cmems_id in ['nextsimre','nextsimanfc']: 
        variables = ["sea_ice_area_fraction",]
    elif cmems_id in ['glophyanfcD',]: 
        variables = ["sea_ice_area_fraction",
                "sea_ice_x_velocity",
                "sea_ice_y_velocity"]
    else:
        variables = ['x_sea_water_velocity',
                'y_sea_water_velocity',
                "sea_ice_area_fraction",
                "sea_ice_x_velocity",
                "sea_ice_y_velocity"]
        
    # cmems_id = env_dict[cmems_id]
    print(cmems_id,variables)
    
    if str(env_dict[cmems_id]).startswith('cmems_'):
    
        from opendrift.readers import reader_copernicusmarine
        reader = reader_copernicusmarine.Reader(
            dataset_id=env_dict[cmems_id])
    elif cmems_id=='topaz5-ensemble':
        ds_env = xr.open_mfdataset(env_dict[cmems_id],
                    concat_dim=xr.DataArray(range(len(env_dict[cmems_id])), dims='member', name='member',attrs={'standard_name': 'realization'}),
                    combine='nested', data_vars='all', coords='all', chunks={'time': 1}) #Solution from KF!
        from opendrift.readers import reader_netCDF_CF_generic
        reader = reader_netCDF_CF_generic.Reader(ds_env)
        del ds_env
    else:
        from opendrift.readers import reader_netCDF_CF_generic
        reader = reader_netCDF_CF_generic.Reader(
            env_dict[cmems_id])
        
    try:z0 = min(0.0, np.max(reader.z))
    except: z0=0
    
    for i,t in enumerate(tl):
        # print('%s/%s'%(i,len(tl)))
        env, profiles = reader.get_variables_interpolated(
            variables=variables,
            lon=obs.lon[i],
            lat=obs.lat[i],
            time=np.datetime64(t),
            z=z0
        )
        for v in variables:
            out[v].loc[{'time_obs':t,'cmems_id':cmems_id}] = env[v].data[0]
```

## Calculate Eddy kinetic energy in region

```python
import xarray as xr
obs = xr.open_dataset('/home/jovyan/work/tutorials/sources/OpenDrift/openberg_july2026/input/iceberg-2026.nc')
```

```python
#READ DATA
cmems_id = env_dict['glophyanfcH']
# vars_cmems = ['vxo','vyo']
vars_cmems = ['uo','vo']
# load_period = slice('2026-06-01','2026-08-30')
load_period = slice(str(obs.time[0].data),str(obs.time[-1].data))
load_period2=slice('2026-07-25','2026-08-10')# or load_period #only affects spatial subset

obs_sub_p = obs.sel(time=load_period2)
# from opendrift.readers import reader_copernicusmarine
# reader = reader_copernicusmarine.Reader(
#     dataset_id=env_dict[cmems_id])
import copernicusmarine as cm
cmems = cm.open_dataset(
            dataset_id=cmems_id,
            variables=vars_cmems,
            minimum_longitude=obs_sub_p.lon.min()-1,
            maximum_longitude=obs_sub_p.lon.max()+1,
            minimum_latitude=obs_sub_p.lat.min()-0.5,
            maximum_latitude=obs_sub_p.lat.max()+0.5,
            start_datetime=load_period.start,#str(obstime[0].values),
            end_datetime=load_period.stop,#str(obstime[1].values),
        )

if 'depth' in cmems.dims: cmems=cmems.sel(depth=cmems.depth[0].data)
```

```python
u_mean = cmems[vars_cmems[0]].mean(axis=0)
v_mean = cmems[vars_cmems[1]].mean(axis=0)
# v_bg = cmems.vyo.rolling(time=31, center=True).mean()

u_prime = (cmems[vars_cmems[0]]-u_mean)
v_prime = (cmems[vars_cmems[1]]-v_mean)

mke = 0.5*1025*(u_mean**2+v_mean**2)
eke = 0.5*1025*(u_prime**2+v_prime**2)
eke_mean= eke.mean(axis=0)
```

```python
# np.hypot(cmems.vxo[0],cmems.vyo[0]).plot()
# eke.mean(axis=0).plot()
```

```python
lat1,lon1,v1 = cmems.latitude.data,cmems.longitude.data,np.hypot(cmems[vars_cmems[0]][0],cmems[vars_cmems[1]][0])
vxo_date = '2026-08-05T00'

#Figure
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
from matplotlib.patches import Patch
from matplotlib.lines import Line2D
fig,ax = plt.subplots(1,1,figsize=(10,10),subplot_kw={'projection':ccrs.NorthPolarStereo(central_longitude=float(cmems.longitude.mean()))})#central_longitude=20})
ax.coastlines(resolution='10m')
trans = ccrs.PlateCarree()
#plot contours
# cf = ax.contourf(lon1,lat1,mke,transform=trans)
cf = ax.contourf(lon1,lat1,eke_mean,transform=trans)
plt.colorbar(cf,shrink=0.6)
#plot currents
dd=1
ax.quiver(lon1[::dd],lat1[::dd],cmems[vars_cmems[0]].sel(time=vxo_date).data[::dd,::dd],cmems[vars_cmems[1]].sel(time=vxo_date).data[::dd,::dd],transform=trans)
#plot obs
ax.plot(obs['lon'].sel(time=load_period2),obs['lat'].sel(time=load_period2),'-',transform=trans,color='k',label='Iceberg trajectory')
ax.plot(obs['lon'].sel(time=load_period2),obs['lat'].sel(time=load_period2),'^',transform=trans,color='k',label='Individual observations')

h, l = ax.get_legend_handles_labels()
by_label = dict(zip(l, h))
# by_label['TOPAZ5 current (%s)'%vxo_date] = Line2D([], [],color='k',marker='>',markersize=10,linestyle='-')
by_label['GLOBAL current (%s)'%vxo_date] = Line2D([], [],color='k',marker='>',markersize=10,linestyle='-')
by_label['Eddy kinetic energy'] = Patch( color='darkblue')
ax.legend(handles=list(by_label.values())[::-1],
         labels=list(by_label.keys())[::-1])

note,env='_P2','GLO12'
# fig.savefig('./results/%s_map_%s_vxo_eddy%s.png'%(ib,env,note),dpi=200,transparent=True,bbox_inches='tight')
plt.show()
```
