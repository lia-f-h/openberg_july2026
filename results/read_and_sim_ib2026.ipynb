```python
#read data
df2 = pd.read_csv('/home/jovyan/work/tutorials/sources/OpenDrift/openberg_july2026/input/iceberg_202609_breaks.csv')
df = pd.read_csv('/home/jovyan/work/tutorials/sources/OpenDrift/openberg_july2026/input/iceberg-2026_all.csv')
```

```python
#fix df
timestr = [str(t) if t>99999 else '0'+str(t) for t in df['Time']]
df["time"] = pd.to_datetime(
    df["Date"].astype(str) + " " + timestr,
    utc=True
).dt.tz_convert(None)
df = df.sort_values('id')

df = df[['Date','time', 'lat', 'lon']]
df
```

```python
#fix df
timestr = [str(t) if t>99999 else '0'+str(t) for t in df2['Time']]
df2["time"] = pd.to_datetime(
    df2["Date"].astype(str) + " " + timestr,
    utc=True
).dt.tz_convert(None)
df2 = df2.sort_values('id')
df2['lat'],df2['lon']=df2['longitude'],df2['Latitude']


df2 = df2[['Date','time', 'lon', 'lat']]
# df2
```

```python
df3 = pd.concat([df, df2], ignore_index=True)
ds=df3.set_index("time").sort_values('time').to_xarray()
ds.to_netcdf('/home/jovyan/work/tutorials/sources/OpenDrift/openberg_july2026/input/iceberg-2026.nc')
```

```python
import json

date = '2026/07/29'#28
date2 = pd.to_datetime('2026-08-07')
df_date = df.loc[df.Date==date]
leadt=date2-df_date.time.values[0]
leadt=leadt.days+1

iceberg = {
    "time": str(df_date.time.values[0]),
    "lat": float(df_date.lat.values[0]),
    "lon": float(df_date.lon.values[0]),
    "length": 5500,
    "width": 4000,
    "draft": 200,
    "sail": 25,
}

cmd = (
    "python3 -m openberg_july2026.scripts.sim_argparse "
    f"--argib 'iceberg2026e' "
    f"--argobs 'None' "
    f"--argiceberg '{json.dumps(iceberg, separators=(',', ':'))}' "
    f"--argoc '[[\"gebco\",\"topaz5\"]]' "
    f"--argleadtime '{leadt}' "
    "--argmainrun 1 "
    "--argname 'ib%s_test'"%(date[5:7]+date[8:])
)

print(cmd)
```
