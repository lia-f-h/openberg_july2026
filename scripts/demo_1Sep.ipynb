# Simulating a large iceberg in the Greenland Sea

```python
from opendrift.models.openberg import OpenBerg
from datetime import datetime
import numpy as np
```

```python
o=OpenBerg(loglevel=10,logfile='log1.out')
```

```python
o.add_readers_from_list(['cmems_mod_arc_phy_anfc_6km_detided_PT1H-i',
                        'cmems_obs-wind_glo_phy_nrt_l4_0.125deg_PT1H']) 
```

```python
o.seed_elements(
                #iceberg size
                length=5000, width=3000, draft=150, sail=11,
                #iceberg time, position
                lon=-24.71,
                lat=68.457,
                time=datetime(2026,7,1,8,21,9),
                #for seeding multiple icebergs
                number=1,
                radius=0
                )
```

```python
run_dict = {
    # 'duration':timedelta(days=2),
    'end_time':datetime(2026,7,10),
    'outfile':'resulst.nc'
}
o1 = o.run(**run_dict)
```
